```python
df_10k = pd.read_csv('10k_sample.csv')
df_10k
```

```python
# reusing code from class08 notebooks to clean text
import string
translator = str.maketrans('', '', string.punctuation)
df_10k['item1'] = df_10k['item_1_text'].apply(lambda s: s.lower().translate(translator))
df_10k['wc'] = df_10k['item1'].apply(lambda s: len(s.split()))
df_10k[df_10k['wc'] == df_10k['wc'].max()]
```

```python
df_10k['slsts'] = df_10k['item_1_text'].apply(lambda s: s.split('.'))
df_10k['avg_slen'] = df_10k['slsts'].apply(lambda lsts: sum(len(s) for s in lsts)/len(lsts))
round(df_10k['avg_slen'].mean())
```

```python
# reusing code from class08 notebooks to clean text
import nltk
nltk.download('stopwords')
eng_stopwords = set(nltk.corpus.stopwords.words('english'))
df_10k['item1c'] = df_10k['item1'].apply(lambda s: [w for w in s.split() if w not in eng_stopwords])
df_10k['wc1'] = df_10k['item1c'].apply(lambda lst: len(lst))
df_10k[df_10k['wc1'] == df_10k['wc1'].min()]
```

```python
df_10k['bwp'] = df_10k['item1c'].apply(lambda lst: sum(len(w) >= 7 for w in lst) / len(lst))
df_10k[df_10k['bwp'] == df_10k['bwp'].max()]
```

```python
# How to check if col val contains a substring
# ref: https://stackoverflow.com/questions/30944577/check-if-string-is-in-a-pandas-dataframe
df_10k['item1s'] = df_10k['item1c'].apply(lambda lst: ' '.join(lst))
df_10k['ai_count'] = df_10k['item1c'].apply(lambda lst: sum(w == 'ai' for w in lst)) 
(df_10k['ai_count'] > 0 | df_10k['item1s'].str.contains('deep learning')).sum()
```

```python
crime = pd.read_csv('crime.csv')
crime
```

```python
crime_wpg = crime[crime['NEIGHBOURHOOD'] == 'West Point Grey']
# ref: https://stackoverflow.com/questions/19384532
crime_wpg.groupby('TYPE').size()
# From inspection below, the second most common is 'Break and Enter Residential/Other'
```

```python
crime[crime['TYPE'] == 'Theft of Vehicle'].groupby('YEAR').size().nsmallest(2)
```

```python
day_cond = (crime['HOUR'] >= 7) & (crime['HOUR'] < 23)
night_cond = (crime['HOUR'] < 7) | (crime['HOUR'] >= 23)
```

```python
fatal_cond = crime['TYPE'] == 'Vehicle Collision or Pedestrian Struck (with Fatality)'
acc_cond = fatal_cond | (crime['TYPE'] == 'Vehicle Collision or Pedestrian Struck (with Injury)')
```

```python
crime_day_fatal = crime[day_cond & fatal_cond]
crime_day_acc = crime[day_cond & acc_cond]
crime_day_rate = len(crime_day_fatal) / len(crime_day_acc)
crime_day_rate
```

```python
crime_night_fatal = crime[night_cond & fatal_cond]
crime_night_acc = crime[night_cond & acc_cond]
crime_night_rate = len(crime_night_fatal) / len(crime_night_acc)
crime_night_rate
```

```python
(crime_night_rate - crime_day_rate) * 100.0 / crime_day_rate
```
