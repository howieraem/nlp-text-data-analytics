```python
import numpy as np
import pandas as pd
from scipy.spatial.distance import cityblock, euclidean, cosine
from sklearn.metrics.pairwise import cosine_similarity
import matplotlib.pyplot as plt
from scipy.spatial.distance import pdist, squareform
```

# **1. Minkowski Distance (Manhattan and Euclidean)**

```python
# Create a small dataset for similarity calculations
data1 = pd.DataFrame({
    'Movie.1': [1, 5],
    'Movie.2': [2, 3]
}, index=['Alice', 'Bernard'])

data1
```

## `pdist` - Pairwise Distance in SciPy

`pdist` stands for **pairwise distance** and is a function from SciPy's `spatial.distance` module. It computes the **distance between every pair of rows** in a dataset.

### Parameters:
- **`data`**: A matrix where each row is a point (observation).
- **`metric`**: The type of distance to compute (e.g., `'euclidean'`, `'cityblock'`, `'cosine'`, etc.).


```python
# Calculate Manhattan Distance (L1 norm)
pdist(data1, metric='cityblock')
```

We can also produce a matrix, this becomes useful when we have more than 2 users

```python
# square matrix
manhattan_dist_matrix_1 = pd.DataFrame(squareform(pdist(data1, metric='cityblock')),
                                     index=data1.index,
                                     columns=data1.index)
manhattan_dist_matrix_1
```

```python
# Calculate Euclidean Distance (L2 norm)
pdist(data1, metric='euclidean')
```

```python
# Compute Euclidean Distance (L2 Norm)
euclidean_dist_matrix_1 = pd.DataFrame(squareform(pdist(data1, metric='euclidean')),
                                     index=data1.index,
                                     columns=data1.index)
euclidean_dist_matrix_1
```

# **2. Correlation Calculation**

```python
data1.T.corr(method='pearson')
```

The .corr() function in pandas computes the correlation matrix between numerical columns in a DataFrame. It returns a square matrix where each cell represents the correlation between two columns.

### Diagonal Values (1.0):
- Alice's and Bernard's ratings are perfectly correlated with themselves (by definition).

### Off-Diagonal Values (-1.0):
- Indicates a perfect negative correlation between Alice and Bernard, meaning: Alice likes movie 2 better while Bernard likes movie 1 better



# **3. Cosine Distance**

```python
# can also use pdist, note this is distance
cos_dist_matrix_1 = pd.DataFrame(squareform(pdist(data1, metric='cosine')),
                                     index=data1.index,
                                     columns=data1.index)
cos_dist_matrix_1
```

```python
# can also use cosine_similarity, this is calculating cosine_similarity, which is 1-cosine_distance
cosine_similarity(data1)
```

# **4. Handling Missing Data**

```python
data2 = pd.DataFrame({
    'Movie.1': [np.nan, 1],
    'Movie.2': [np.nan, 2],
    'Movie.3': [4, 2],
    'Movie.4': [4, 2],
    'Movie.5': [5, 3]
}, index=['Alice', 'Bernard'])

print("Dataset 2 with missing values:")
print(data2)
```

## Pearson Distane

The default option is to ignore missing values when calculating correlation.

```python
# Compute pearson distance
pearson_dist2 = data2.T.corr(method='pearson')
pearson_dist2
```

## Cosine distance

```python
# cosine distance: ignoring missing values, only compare the ratings when both users rate
# Remove columns with any missing values
data2_cleaned = data2.dropna(axis=1, how='any')
```

```python
data2_cleaned
```

```python
cos_dist_matrix_2 = pd.DataFrame(squareform(pdist(data2_cleaned, metric='cosine')),
                                     index=data2_cleaned.index,
                                     columns=data2_cleaned.index)
cos_dist_matrix_2
```

#Compare distance functions after scaling one set of values

Cindy's scores are Alice's plus a constant

```python
data3 = pd.DataFrame({
    'Movie.1': [2, 1, 7],
    'Movie.2': [2, 2, 7],
    'Movie.3': [4, 2, 9],
    'Movie.4': [4, 2, 9],
    'Movie.5': [5, 3, 10]
}, index=['Alice', 'Bernard', 'Cindy'])

print("Dataset with scaled values:")
print(data3)
```

```python

manhattan_dist_matrix_3 = pd.DataFrame(squareform(pdist(data3, metric="cityblock")),
                                     index=data3.index,
                                     columns=data3.index)

manhattan_dist_matrix_3
```

Manhattan distance indicates the distance between Alice and Cindy is 25.

```python
#if you dont use squareform it becomes less intuitive
pdist(data3, metric="cityblock")
```

```python
# Compute Euclidean Distance (L2 Norm)
euclidean_dist_matrix_3 = pd.DataFrame(squareform(pdist(data3, metric="euclidean")),
                                     index=data3.index,
                                     columns=data3.index)
euclidean_dist_matrix_3
```

Euclidean distance indicates the distance between Alice and Cindy is 11.18
